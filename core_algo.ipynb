```python
inventory_q = 0
risk_factor = 0.1
# t_remaining = 1 # For continuous markets
kappa = 1.5

price_ask_list = []
price_bid_list = []

total_time = (df.iloc[-1].event_time - df.iloc[0].event_time)

spot_prices = df["spot_price"].values.tolist()
```

```python
for i in range(1001, 2000):
    current_spot_price = spot_prices[i]
    hund_values = spot_prices[i - 1000 - 1: i]

    normalized_window = (hund_values / current_spot_price) * 100

    current_spot_norm = 100.0
    
    variance = np.var(normalized_window)
    t_remaining = (total_time - df.iloc[i].event_time)/total_time
    reservation_price = current_spot_norm - (inventory_q * risk_factor * variance * t_remaining)
    delta = (1/ 2) * risk_factor * variance * t_remaining + (1/risk_factor)* np.log(1 + (risk_factor/kappa))
    
    price_ask_norm = reservation_price + delta
    price_bid_norm = reservation_price - delta
    
    price_ask = (price_ask_norm / 100.0) * current_spot_price
    price_bid = (price_bid_norm / 100.0) * current_spot_price
    
    price_ask_list.append(price_ask)
    price_bid_list.append(price_bid)
```
